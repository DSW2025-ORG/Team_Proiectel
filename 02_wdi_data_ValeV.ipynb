```python
# ---------------------------------------------------------
# Descarcăm indicatorii WDI într-un DataFrame
# ---------------------------------------------------------
# -----------------------------------------
def get_wdi(indicator_code):
    url = f"https://api.worldbank.org/v2/country/all/indicator/{indicator_code}"
    params = {"format": "json", "per_page": 20000}
    return requests.get(url, params=params).json()

# ---------------------------------------------------------
# Lista finală de indicatori (WDI_INDICATORS)
# ---------------------------------------------------------
WDI_INDICATORS = {

     # -----------------------------------------
    # A) SOCIO-ECONOMIC CORE
    # -----------------------------------------
    "gdp_per_capita": "NY.GDP.PCAP.CD",        # Nivelul de dezvoltare economică (capacitate investiții)
    "population": "SP.POP.TOTL",               # Bază pentru CO₂/capita și energie/capita
    "urbanization_rate": "SP.URB.TOTL.IN.ZS",  # Urbanizare — cerere energetică, modernizare structurală

    # -----------------------------------------
    # B) ENERGY CONTEXT (complement to EIA)
    # -----------------------------------------
    "energy_use_per_capita": "EG.USE.PCAP.KG.OE",   # Stil energetic, consum structural
    "renewable_final_share": "EG.FEC.RNEW.ZS",      # Regenerabile în consum final (complement EIA)
    "electricity_access": "EG.ELC.ACCS.ZS",         # Acces la electricitate (energy poverty vs transition)
    "investment_energy": "IE.PPI.ENGY.CD",         # Investment in energy with private participation (current US$)

    # -----------------------------------------
    # C) GOVERNANCE (WGI Landscape)
    # -----------------------------------------
    "gov_effectiveness": "GE.EST",       # Capacitatea statului de a implementa tranziții
    "regulatory_quality": "RQ.EST",      # Calitatea politicilor (energia este sector reglementat)
    "rule_of_law": "RL.EST",             # Stabilitate instituțională
    "control_corruption": "CC.EST",      # Integritate instituțională (investiții în infrastructură)


    # -----------------------------------------
    # E) EMISSIONS / GREENHOUSE GAS
    # -----------------------------------------
    "CO2_emissions_proc": "EN.GHG.CO2.ZG.AR5",  # Carbon dioxide (CO2) emissions (total) excluding LULUCF (% change from 1990)
    "CO2_emissions_mt": "EN.GHG.CO2.MT.CE.AR5",  # 	Carbon dioxide (CO2) emissions (total) excluding LULUCF (Mt CO2e)
    "N2O_emissions_proc": "EN.GHG.N2O.ZG.AR5",  # Nitrous oxide (N2O) emissions (total) excluding LULUCF (% change from 1990)
    "N2O_emissions_mt": "EN.GHG.N2O.MT.CE.AR5",  # 	Nitrous oxide (N2O) emissions (total) excluding LULUCF (Mt CO2e)
    "CH4_emissions_proc": "EN.GHG.CH4.ZG.AR5",  # Methane (CH4) emissions (total) excluding LULUCF (% change from 1990)
    "CH4_emissions_mt": "EN.GHG.CH4.MT.CE.AR5"
}

# ---------------------------------------------------------
# Descărcăm toți indicatorii într-un dicționar de DataFrame-uri
# ---------------------------------------------------------

wdi_data = {}

for name, code in WDI_INDICATORS.items():
    print(f"Downloading {name} ({code}) ...")
    wdi_data[name] = get_wdi(code)


```

```python
import pandas as pd
from functools import reduce

# Funcție de transformare JSON WDI → DataFrame
def wdi_json_to_df(wdi_json, value_name):
    """
    wdi_json: lista cu 2 elemente [metadata, data_list]
    value_name: numele coloanei finale pentru indicator
    """
    # selectăm doar lista cu observații (index 1)
    data_list = wdi_json[1] if len(wdi_json) > 1 else []
    if not data_list:
        return pd.DataFrame(columns=["country_code", "country", "year", value_name])
    
    # normalizăm JSON-ul
    df = pd.json_normalize(data_list)

    # redenumim coloanele
    df = df.rename(columns={
        "country.id": "country_code",
        "country.value": "country",
        "date": "year",
        "value": value_name
    })

    # convertim tipurile
    df["year"] = pd.to_numeric(df["year"], errors="coerce").astype("Int64")
    df[value_name] = pd.to_numeric(df[value_name], errors="coerce")

    # păstrăm doar coloanele relevante
    return df[["country_code", "country", "year", value_name]]

# -------------------------
# Cream DataFrame pentru fiecare indicator
# -------------------------
dfs = [
    wdi_json_to_df(gdp_pc_json, "gdp_per_capita"),
    wdi_json_to_df(population_json, "population"),
    wdi_json_to_df(urbanization_rate_json, "urbanization_rate"),
    wdi_json_to_df(energy_use_per_capita_json, "energy_use_per_capita"),
    wdi_json_to_df(renewable_final_share_json, "renewable_final_share"),
    wdi_json_to_df(electricity_access_json, "electricity_access"),
    wdi_json_to_df(investment_energy_json, "investment_energy"),
    wdi_json_to_df(gov_effectiveness_json, "gov_effectiveness"),
    wdi_json_to_df(regulatory_quality_json, "regulatory_quality"),
    wdi_json_to_df(rule_of_law_json, "rule_of_law"),
    wdi_json_to_df(control_corruption_json, "control_corruption"),
    wdi_json_to_df(CO2_emissions_proc_json, "CO2_emissions_proc"),
    wdi_json_to_df(CO2_emissions_mt_json, "CO2_emissions_mt"),
    wdi_json_to_df(N2O_emissions_proc_json, "N2O_emissions_proc"),
    wdi_json_to_df(N2O_emissions_mt_json, "N2O_emissions_mt"),
    wdi_json_to_df(CH4_emissions_proc_json, "CH4_emissions_proc"),
    wdi_json_to_df(CH4_emissions_mt_json, "CH4_emissions_mt")
]

# -------------------------
# Unim toate DataFrame-urile pe country_code + country + year
# -------------------------
df = reduce(
    lambda left, right: pd.merge(
        left, right,
        on=["country_code", "country", "year"],
        how="outer"
    ),
    dfs
)

# -------------------------
# Verificăm tipurile și primele rânduri
# -------------------------
print(df.dtypes)
print(df.head())
```
